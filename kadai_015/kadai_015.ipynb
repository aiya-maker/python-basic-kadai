```python
# クラスの定義（例：class Product: ...... ）
class Human:
    # コンストラクタの定義
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def printinfo(self):
      print(self.name)
      print(self.age)

# インスタンス化（例：shampoo = Product()）
human = Human("花子", 30)

human.printinfo()
```
